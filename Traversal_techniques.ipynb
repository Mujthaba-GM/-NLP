```python
# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# Starting node
queue = ['A']

# Visited nodes
visited = []

# BFS
while queue:

    # Take the first node
    node = queue.pop(0)

    # Visit the node
    visited.append(node)

    # Add its neighbours to queue
    for n in graph[node]:
        queue.append(n)

# Print BFS order
print(visited)
```

```python
#DFS
# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': [],
    'D': [],
    'E': []
}

# Stack
stack = ['A']

# Visited nodes
visited = []

# DFS
while stack:

    # Take the top node
    node = stack.pop()

    # Visit the node
    if node not in visited:
        visited.append(node)
        print("Visit:", node)

        # Add neighbours
        for n in graph[node]:
            if n not in visited:
                stack.append(n)

print("DFS:", visited)
```
