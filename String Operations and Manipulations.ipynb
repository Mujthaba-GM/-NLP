```python
# ==============================
# ALL STRING OPERATIONS IN PYTHON
# ==============================
text = "Natural Language Processing with Python 123!"
print("Original String :", text)
# 1. Lowercase
print("\nLowercase :", text.lower())
# 2. Uppercase
print("Uppercase :", text.upper())
# 3. Capitalize
print("Capitalize :", text.capitalize())
# 4. Title Case
print("Title :", text.title())

# 5. Swap Case
print("Swapcase :", text.swapcase())
# 6. Length
print("Length :", len(text))
# 7. Replace
print("Replace :", text.replace("Python", "NLP"))
# 8. Find
print("Find 'Language' :", text.find("Language"))
# 9. Count
print("Count of 'a' :", text.count("a"))
# 10. Startswith
print("Starts with 'Natural' :", text.startswith("Natural"))
# 11. Endswith
print("Ends with '!' :", text.endswith("!"))
# 12. Split
words = text.split()
print("Split :", words)
# 13. Join
joined = "-".join(words)
print("Join :", joined)
# 14. Strip
text2 = "   Hello NLP   "
print("Strip :", text2.strip())

# 15. Left Strip
print("LStrip :", text2.lstrip())

# 16. Right Strip
print("RStrip :", text2.rstrip())

# 17. Index
print("Index of 'Processing' :", text.index("Processing"))

# 18. Is Alphabet
print("'Python'.isalpha() :", "Python".isalpha())

# 19. Is Digit
print("'123'.isdigit() :", "123".isdigit())

# 20. Is Alphanumeric
print("'Python123'.isalnum() :", "Python123".isalnum())

# 21. Is Lower
print("'python'.islower() :", "python".islower())

# 22. Is Upper
print("'PYTHON'.isupper() :", "PYTHON".isupper())

# 23. Is Space
print("'   '.isspace() :", "   ".isspace())

# 24. Reverse String
print("Reverse :", text[::-1])

# 25. Remove Numbers
import re
print("Remove Numbers :", re.sub(r'\d+', '', text))

# 26. Remove Punctuation
import string
clean = text.translate(str.maketrans('', '', string.punctuation))
print("Remove Punctuation :", clean)

# 27. Tokenization
tokens = clean.split()
print("Tokens :", tokens)

# 28. Character Count
print("Characters :", len(clean))

# 29. Word Count
print("Words :", len(tokens))

# 30. Concatenation
s1 = "Natural"
s2 = "Language"
print("Concatenation :", s1 + " " + s2)

# 31. Repetition
print("Repetition :", "NLP " * 3)

# 32. String Slicing
print("First 7 Characters :", text[:7])
print("Last 6 Characters :", text[-6:])

# 33. Membership
print("'Python' in text :", "Python" in text)

# 34. ASCII Encoding
print("Encode :", text.encode())

# 35. Formatting
name = "Alice"
course = "NLP"
print("Formatted String :", f"{name} is learning {course}")

# 36. Center
print("Center :", "Python".center(20, '-'))

# 37. Left Justify
print("Ljust :", "Python".ljust(20, '-'))

# 38. Right Justify
print("Rjust :", "Python".rjust(20, '-'))

# 39. Zero Fill
print("Zfill :", "25".zfill(5))

# 40. Partition
print("Partition :", text.partition("Language"))
```
