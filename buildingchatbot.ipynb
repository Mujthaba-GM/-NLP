```python
#Load spacy
import spacy

nlp = spacy.load("en_core_web_sm")

print("NLP model loaded successfully")
```

```python
#Create Healthcare Knowledge
answers = {
    "fever": "Fever may include increased body temperature, sweating, chills and weakness.",

    "headache": "Headache can have many causes. Persistent or severe headache should be evaluated by a healthcare professional.",

    "cold": "Common cold symptoms may include sneezing, runny nose, sore throat and cough.",

    "cough": "Cough can have many causes. Persistent or severe cough should be evaluated by a healthcare professional.",

    "diabetes": "Diabetes is a condition involving abnormal blood glucose levels and requires professional medical evaluation.",

    "blood pressure": "Blood pressure measures the force of blood against artery walls. Abnormal readings should be discussed with a healthcare professional."
}
```

```python
#create the chat bot
def chatbot(question):

    text = question.lower()

    for key in answers:
        if key in text:
            return answers[key]

    return "Sorry, I don't understand the question."
```

```python
#Test the Chatbot
question = input("You: ")

answer = chatbot(question)

print("Chatbot:", answer)
```

```python
#Make It a Continuous Chatbot
print("Healthcare Chatbot")
print("Type 'exit' to stop")

while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        print("Bot: Thank you!")
        break

    answer = chatbot(question)

    print("Bot:", answer)
```

```python
#Show the NLP Operations
question = input("Enter your question: ")

doc = nlp(question)

print("\nTOKENIZATION")
for token in doc:
    print(token.text)

print("\nPOS TAGGING")
for token in doc:
    print(token.text, "→", token.pos_)

print("\nLEMMATIZATION")
for token in doc:
    print(token.text, "→", token.lemma_)
```

```python
#Add NLP Processing
def process_text(question):

    doc = nlp(question.lower())

    print("\nTokens:")

    for token in doc:
        print(token.text)

    print("\nPOS Tags:")

    for token in doc:
        print(token.text, "→", token.pos_)

    print("\nLemmas:")

    for token in doc:
        print(token.text, "→", token.lemma_)

    return doc
```

```python
question = input("Enter a question: ")

process_text(question)
```

```python
#Combine NLP+Chatbot
def chatbot(question):

    doc = nlp(question.lower())

    text = question.lower()

    print("\nTokens:")

    for token in doc:
        print(token.text, end=" | ")

    print("\n")

    for key in answers:

        if key in text:
            return answers[key]

    return "Sorry, I do not understand the question."
```

```python
question = input("You: ")

print("Chatbot:", chatbot(question))
```
