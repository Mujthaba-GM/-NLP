```python
#tokenization
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I am learning Natural Language Processing."

doc = nlp(text)

for token in doc:
    print(token.text)
```

```python
#Sentence Tokenization
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I am learning NLP. spaCy is very useful."

doc = nlp(text)

for sent in doc.sents:
    print(sent.text)
```

```python
#POS Tagging
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The boy is playing cricket."

doc = nlp(text)

for token in doc:
    print(token.text, token.pos_)
```

```python
#Detailed POS Tagging
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The small boy is playing football."

doc = nlp(text)

for token in doc:
    print(token.text, token.pos_, token.tag_)
```

```python
#5. Lemmatization
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The boys was playing games."

doc = nlp(text)

for token in doc:
    print(token.text, "->", token.lemma_)
```

```python
#Stop Word Detection
import spacy

nlp = spacy.load("en_core_web_sm")

text = "This is a simple NLP program."

doc = nlp(text)

for token in doc:
    print(token.text, token.is_stop)
```

```python
#NER
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Sundar Pichai is the CEO of Google."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
#9. Display Named Entities
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

text = "Alexa Smith works at Google in California."

doc = nlp(text)

displacy.render(doc, style="ent", jupyter=True)
```

```python
#Dependency Parsing
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The boy plays cricket."

doc = nlp(text)

for token in doc:
    print(token.text, token.dep_, token.head.text)
```

```python
#Word Similarity
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("cat dog car")

print(doc[0].similarity(doc[1]))
print(doc[0].similarity(doc[2]))
```

```python
#Dependency Tree Visualization
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

text = "The boy plays cricket."

doc = nlp(text)

displacy.render(doc, style="dep", jupyter=True)
```

```python
#count words
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I am learning NLP using spaCy."

doc = nlp(text)

print("Number of tokens:", len(doc))
```

```python
#Find Only Nouns
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The intelligent student writes a program."

doc = nlp(text)

for token in doc:
    if token.pos_ == "NOUN":
        print(token.text)
```

```python
#Find Only Verbs
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The student writes and submits the assignment."

doc = nlp(text)

for token in doc:
    if token.pos_ == "VERB":
        print(token.text)
```

```python
#Extract Person Names
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Rahul met Priya at the college."

doc = nlp(text)

for ent in doc.ents:
    if ent.label_ == "PERSON":
        print(ent.text)
```

```python
#. Extract Locations
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Hyderabad is a city in India."

doc = nlp(text)

for ent in doc.ents:
    if ent.label_ == "GPE":
        print(ent.text)
```
