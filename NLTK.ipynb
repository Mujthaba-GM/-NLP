```python
import re
import string

text = "I DON'T like   NLP!!!"

# 1. Lowercase
text = text.lower()

# 2. Expand contraction
text = text.replace("don't", "do not")

# 3. Remove punctuation
text = text.translate(
    str.maketrans('', '', string.punctuation)
)

# 4. Remove extra spaces
text = re.sub(r'\s+', ' ', text).strip()

print(text)
```

```python
import nltk

# Download required data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('punkt_tab') # Added to resolve the LookupError
nltk.download('averaged_perceptron_tagger_eng') # Added to resolve the new LookupError

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk

text = "Ravi is studying Artificial Intelligence in Hyderabad."

print("Original Text:")
print(text)

# 1. Tokenization
words = word_tokenize(text)
print("\n1. Tokens:")
print(words)

# 2. Stop-word removal
stop_words = set(stopwords.words("english"))
new_words = []

for word in words:
    if word.lower() not in stop_words:
        new_words.append(word)

print("\n2. After Stop-word Removal:")
print(new_words)

# 3. Stemming
stemmer = PorterStemmer()

print("\n3. Stemming:")
for word in new_words:
    print(word, "->", stemmer.stem(word))

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()

print("\n4. Lemmatization:")
for word in new_words:
    print(word, "->", lemmatizer.lemmatize(word))

# 5. POS Tagging
print("\n5. POS Tagging:")
print(pos_tag(words))

# 6. Named Entity Recognition
print("\n6. Named Entities:")
tree = ne_chunk(pos_tag(words))
print(tree)
```
