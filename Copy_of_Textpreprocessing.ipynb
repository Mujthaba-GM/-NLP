```python
#Character Classes [abc]
import re

text = input("Enter a string: ")

result = re.findall(r"[abc]", text)

print("Matched Characters:")
print(result)
```

```python
#Negated Character Class [^abc]
import re

text = input("Enter a string: ")

result = re.findall(r"[^abc]", text)

print(result)
```

```python
#Program 3: Lowercase Letters [a-z]
import re

text = input("Enter a string: ")

result = re.findall(r"[a-z]", text)

print(result)
```

```python
#Program 4: Uppercase Letters [A-Z]
import re

text = input("Enter a string: ")

result = re.findall(r"[A-Z]", text)

print(result)
```

```python
#Digits [0-9]
import re

text = input("Enter a string: ")

result = re.findall(r"[0-9]", text)

print(result)
```

```python
#Dot .
import re

text = input("Enter a string: ")

result = re.findall(r"c.t", text)

print(result)
```

```python
#Question Mark
import re

text = input("Enter a string: ")

result = re.findall(r"colou?r", text)

print(result)
```

```python
#Caret ^ (Start of String)
import re

text = input("Enter a string: ")

result = re.findall(r"^The", text)

print(result)
```

```python
#Dollar $ (End of String)
import re

text = input("Enter a string: ")

result = re.findall(r"end$", text)

print(result)
```

```python
#Word Boundary \b
import re

text = input("Enter a string: ")

result = re.findall(r"\bcat\b", text)

print(result)
```

```python
#Exactly n Times {n}
import re

text = input("Enter a string: ")

result = re.findall(r"a{3}", text)

print(result)
```

```python
#n or More {n,}
import re

text = input("Enter a string: ")

result = re.findall(r"b{2,}", text)

print(result)
```

```python
#Between n and m {n,m}
import re

text = input("Enter a string: ")

result = re.findall(r"x{2,4}", text)

print(result)
```

```python
#OR Operator |
import re

text = input("Enter a string: ")

result = re.findall(r"cat|dog", text)

print(result)
```

```python
#1. Tokenization
import re

text = "Natural language processing is fascinating."

tokens = re.findall(r"\w+", text)

print("Tokens:")
print(tokens)
```

```python
#2. Email Extraction
import re

text = "Please contact support@example.com for assistance."

emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)

print("Email Addresses:")
print(emails)
```

```python
#3. URL Extraction
import re

text = "Visit our website at https://www.example.com for more information."

urls = re.findall(r"https?://\S+", text)

print("URLs:")
print(urls)
```

```python
#4. Hashtag Extraction
import re

text = "Join the conversation with #NLP and #MachineLearning."

hashtags = re.findall(r"#\w+", text)

print("Hashtags:")
print(hashtags)
```

```python
#5. Phone Number Extraction
import re

text = "Please call 555-123-4567 for assistance."

phones = re.findall(r"\b\d{3}-\d{3}-\d{4}\b", text)

print("Phone Numbers:")
print(phones)
```

```python
#6. Date Extraction
import re

text = "The event is scheduled for 20/09/2026."

dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", text)

print("Dates:")
print(dates)
```

```python
#7. Mention Extraction
import re

text = "I follow @OpenAI and @PythonHub."

mentions = re.findall(r"@\w+", text)

print("Mentions:")
print(mentions)
```
