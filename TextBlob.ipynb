```python
#Word Tokenization
!python -m textblob.download_corpora
from textblob import TextBlob

blob = TextBlob("The boy plays cricket.")

print(blob.words)
```

```python
#Sentence Tokenization
print(blob.sentences)
```

```python
#POS Tagging
from textblob import TextBlob

blob = TextBlob("The boy plays cricket.")

print(blob.tags)
```

```python
#Noun Phrase Extraction
blob = TextBlob("The intelligent student is studying artificial intelligence.")

print(blob.noun_phrases)
```

```python
#Sentiment Analysis
blob = TextBlob("I really like this movie.")

print(blob.sentiment)
```

```python
#Simple Positive/Negative Sentiment Program
from textblob import TextBlob

text = input("Enter a sentence: ")

blob = TextBlob(text)

p = blob.sentiment.polarity

if p > 0:
    print("Positive")
elif p < 0:
    print("Negative")
else:
    print("Neutral")
```
